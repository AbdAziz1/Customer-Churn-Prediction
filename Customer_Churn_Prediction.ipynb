```python
# Import machine learning models

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier

# Import evaluation metrics

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    classification_report,
    confusion_matrix
)

print("All model libraries imported successfully!")
```

Model Evaluation

```python
# Store predictions and churn probabilities

model_predictions = {}
model_probabilities = {}

for name, model in models.items():

    # Predict customer churn (0 or 1)
    model_predictions[name] = model.predict(X_test_processed)

    # Predict probability of churn
    model_probabilities[name] = model.predict_proba(X_test_processed)

    print(f"{name}: Predictions generated successfully!")

print("\nAll model predictions completed!")
```

```python
# Verify target encoding and model output

print("Target classes:", models["Logistic Regression"].classes_)

print("\nFirst 10 actual values:")
print(y_test.head(10).tolist())

print("\nFirst 10 predictions:")
print(model_predictions["Logistic Regression"][:10])

print("\nFirst 5 probability predictions:")
print(model_probabilities["Logistic Regression"][:5])
```

```python
import pandas as pd

# Create an empty list to store evaluation results
evaluation_results = []

# Evaluate each trained model
for name, model in models.items():

    y_pred = model_predictions[name]

    # Extract probability of class 1 (Churn)
    churn_index = list(model.classes_).index(1)

    y_prob = model_probabilities[name][:, churn_index]

    # Calculate evaluation metrics
    accuracy = accuracy_score(y_test, y_pred)

    precision = precision_score(
        y_test, y_pred, zero_division=0
    )

    recall = recall_score(
        y_test, y_pred, zero_division=0
    )

    f1 = f1_score(
        y_test, y_pred, zero_division=0
    )

    roc_auc = roc_auc_score(y_test, y_prob)

    # Store results
    evaluation_results.append({
        "Model": name,
        "Accuracy": accuracy,
        "Precision": precision,
        "Recall": recall,
        "F1-Score": f1,
        "ROC-AUC": roc_auc
    })

# Create comparison DataFrame
evaluation_df = pd.DataFrame(evaluation_results)

# Sort models by ROC-AUC
evaluation_df = evaluation_df.sort_values(
    by="ROC-AUC",
    ascending=False
)

# Display results
print("MODEL PERFORMANCE COMPARISON")

display(
    evaluation_df.style.format({
        "Accuracy": "{:.4f}",
        "Precision": "{:.4f}",
        "Recall": "{:.4f}",
        "F1-Score": "{:.4f}",
        "ROC-AUC": "{:.4f}"
    })
)
```

Class Imbalance Handling – Random OverSampling

```python
from imblearn.over_sampling import RandomOverSampler
from collections import Counter
```

```python
# Initialize Random OverSampler

ros = RandomOverSampler(random_state=42)

# Oversample training data only

X_train_ros, y_train_ros = ros.fit_resample(
    X_train_processed,
    y_train
)

# Compare class distributions

print("Original training distribution:")
print(Counter(y_train))

print("\nOversampled training distribution:")
print(Counter(y_train_ros))

print("\nOriginal training shape:", X_train_processed.shape)

print("Oversampled training shape:", X_train_ros.shape)
```

```python
# Initialize oversampled models

ros_models = {

    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42,
        n_jobs=-1
    ),

    "Gradient Boosting": GradientBoostingClassifier(
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        random_state=42
    )
}

# Train models using oversampled training data

for name, model in ros_models.items():

    model.fit(X_train_ros, y_train_ros)

    print(f"{name} trained successfully!")
```

```python

# ==========================================
# STEP 10.2B: MODEL OPTIMIZATION SETUP
# ==========================================

import numpy as np

from sklearn.base import BaseEstimator, ClassifierMixin
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.utils.class_weight import compute_sample_weight
from sklearn.model_selection import StratifiedKFold, RandomizedSearchCV

from scipy.stats import randint, uniform, loguniform


# 1. Five-fold stratified cross-validation

cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)


# 2. Custom weighted Gradient Boosting classifier
# Class weights are calculated separately inside each
# training fold to prevent validation-fold leakage.

class WeightedGradientBoosting(ClassifierMixin, BaseEstimator):

    def __init__(
        self,
        n_estimators=100,
        learning_rate=0.1,
        max_depth=3,
        min_samples_split=2,
        min_samples_leaf=1,
        subsample=1.0,
        max_features=None,
        random_state=42
    ):
        self.n_estimators = n_estimators
        self.learning_rate = learning_rate
        self.max_depth = max_depth
        self.min_samples_split = min_samples_split
        self.min_samples_leaf = min_samples_leaf
        self.subsample = subsample
        self.max_features = max_features
        self.random_state = random_state

    def fit(self, X, y):

        weights = compute_sample_weight(
            class_weight="balanced",
            y=y
        )

        self.estimator_ = GradientBoostingClassifier(
            n_estimators=self.n_estimators,
            learning_rate=self.learning_rate,
            max_depth=self.max_depth,
            min_samples_split=self.min_samples_split,
            min_samples_leaf=self.min_samples_leaf,
            subsample=self.subsample,
            max_features=self.max_features,
            random_state=self.random_state
        )

        self.estimator_.fit(
            X,
            y,
            sample_weight=weights
        )

        self.classes_ = self.estimator_.classes_

        return self

    def predict(self, X):
        return self.estimator_.predict(X)

    def predict_proba(self, X):
        return self.estimator_.predict_proba(X)


# 3. Logistic Regression pipeline

lr_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("clf", LogisticRegression(
        class_weight="balanced",
        solver="liblinear",
        max_iter=3000,
        random_state=42
    ))
])


# 4. Baseline Gradient Boosting pipeline

gb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("clf", GradientBoostingClassifier(
        random_state=42
    ))
])


# 5. Weighted Gradient Boosting pipeline

wgb_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("clf", WeightedGradientBoosting(
        random_state=42
    ))
])


# 6. Hyperparameter search spaces

lr_params = {
    "clf__C": loguniform(0.001, 100),
    "clf__penalty": ["l1", "l2"]
}


gb_params = {
    "clf__n_estimators": randint(100, 501),
    "clf__learning_rate": loguniform(0.01, 0.2),
    "clf__max_depth": randint(1, 5),
    "clf__min_samples_split": randint(2, 21),
    "clf__min_samples_leaf": randint(1, 11),
    "clf__subsample": uniform(0.7, 0.3),
    "clf__max_features": [None, "sqrt", "log2"]
}


wgb_params = {
    "clf__n_estimators": randint(100, 501),
    "clf__learning_rate": loguniform(0.01, 0.2),
    "clf__max_depth": randint(1, 5),
    "clf__min_samples_split": randint(2, 21),
    "clf__min_samples_leaf": randint(1, 11),
    "clf__subsample": uniform(0.7, 0.3),
    "clf__max_features": [None, "sqrt", "log2"]
}


# 7. Confirm configuration

print("CROSS-VALIDATION CONFIGURATION")
print(cv)

print("\nOPTIMIZATION CANDIDATES")
print("1. Weighted Logistic Regression")
print("2. Baseline Gradient Boosting")
print("3. Weighted Gradient Boosting")

print("\nSEARCH SPACES")
print("Logistic Regression:", len(lr_params), "parameters")
print("Gradient Boosting:", len(gb_params), "parameters")
print("Weighted Gradient Boosting:", len(wgb_params), "parameters")

print("\nAll optimization pipelines configured successfully.")
```

```python

# ==========================================
# STEP 10.2C: HYPERPARAMETER TUNING
# ==========================================

# 1. Configure randomized searches

lr_search = RandomizedSearchCV(
    estimator=lr_pipeline,
    param_distributions=lr_params,
    n_iter=25,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)

gb_search = RandomizedSearchCV(
    estimator=gb_pipeline,
    param_distributions=gb_params,
    n_iter=25,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)

wgb_search = RandomizedSearchCV(
    estimator=wgb_pipeline,
    param_distributions=wgb_params,
    n_iter=25,
    scoring="f1",
    cv=cv,
    random_state=42,
    n_jobs=-1,
    verbose=1,
    refit=True
)


# 2. Train using training data only

print("TUNING WEIGHTED LOGISTIC REGRESSION")

lr_search.fit(X_train, y_train)

print("\nTUNING BASELINE GRADIENT BOOSTING")

gb_search.fit(X_train, y_train)

print("\nTUNING WEIGHTED GRADIENT BOOSTING")

wgb_search.fit(X_train, y_train)


# 3. Store best estimators

best_lr = lr_search.best_estimator_
best_gb = gb_search.best_estimator_
best_wgb = wgb_search.best_estimator_


# 4. Display optimization results

print("\n" + "=" * 55)
print("HYPERPARAMETER OPTIMIZATION RESULTS")
print("=" * 55)

searches = {
    "Weighted Logistic Regression": lr_search,
    "Baseline Gradient Boosting": gb_search,
    "Weighted Gradient Boosting": wgb_search
}

for name, search in searches.items():

    print(f"\n{name}")
    print("-" * 45)

    print("Best CV F1-score:",
          round(search.best_score_, 4))

    print("Best parameters:")
    print(search.best_params_)
```

```python

# ==========================================
# STEP 10.3A: MULTI-METRIC MODEL COMPARISON
# ==========================================

import pandas as pd

from sklearn.model_selection import cross_val_predict
from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score
)

optimized_models = {
    "Weighted Logistic Regression": best_lr,
    "Baseline Gradient Boosting": best_gb,
    "Weighted Gradient Boosting": best_wgb
}

comparison_results = []

oof_predictions = {}
oof_probabilities = {}

for name, estimator in optimized_models.items():

    print(f"Evaluating: {name}")

    # Out-of-fold class predictions
    predictions = cross_val_predict(
        estimator,
        X_train,
        y_train,
        cv=cv,
        method="predict",
        n_jobs=-1
    )

    # Out-of-fold churn probabilities
    probabilities = cross_val_predict(
        estimator,
        X_train,
        y_train,
        cv=cv,
        method="predict_proba",
        n_jobs=-1
    )[:, 1]

    oof_predictions[name] = predictions
    oof_probabilities[name] = probabilities

    comparison_results.append({
        "Model": name,
        "Precision": precision_score(y_train, predictions),
        "Recall": recall_score(y_train, predictions),
        "F1-Score": f1_score(y_train, predictions),
        "ROC-AUC": roc_auc_score(y_train, probabilities),
        "PR-AUC": average_precision_score(y_train, probabilities)
    })


# Create comparison table

comparison_df = pd.DataFrame(comparison_results)

comparison_df = comparison_df.sort_values(
    by="F1-Score",
    ascending=False
).reset_index(drop=True)

print("\nOPTIMIZED MODEL COMPARISON")
print("=" * 75)

print(comparison_df.round(4).to_string(index=False))
```

```python

# ==========================================
# STEP 10.4A: DECISION THRESHOLD OPTIMIZATION
# ==========================================

import numpy as np
import pandas as pd

from sklearn.metrics import (
    precision_recall_curve,
    precision_score,
    recall_score,
    f1_score
)

# 1. Get Weighted Gradient Boosting OOF probabilities

y_prob_wgb = oof_probabilities[
    "Weighted Gradient Boosting"
]

# 2. Calculate precision and recall at different thresholds

precision, recall, thresholds = precision_recall_curve(
    y_train,
    y_prob_wgb
)

# 3. Calculate F1-score for each threshold

f1_scores = (
    2 * precision[:-1] * recall[:-1]
    / (precision[:-1] + recall[:-1] + 1e-10)
)

# 4. Identify the optimal threshold

best_index = np.argmax(f1_scores)

optimal_threshold = thresholds[best_index]

# 5. Generate predictions using default threshold

y_pred_default = (
    y_prob_wgb >= 0.50
).astype(int)

# 6. Generate predictions using optimal threshold

y_pred_optimal = (
    y_prob_wgb >= optimal_threshold
).astype(int)

# 7. Compare both thresholds

threshold_results = pd.DataFrame([
    {
        "Threshold": 0.50,
        "Precision": precision_score(
            y_train, y_pred_default
        ),
        "Recall": recall_score(
            y_train, y_pred_default
        ),
        "F1-Score": f1_score(
            y_train, y_pred_default
        )
    },
    {
        "Threshold": optimal_threshold,
        "Precision": precision_score(
            y_train, y_pred_optimal
        ),
        "Recall": recall_score(
            y_train, y_pred_optimal
        ),
        "F1-Score": f1_score(
            y_train, y_pred_optimal
        )
    }
])

print("OPTIMAL THRESHOLD")
print("=" * 45)

print("Optimal threshold:", round(optimal_threshold, 4))
print("Maximum CV-based F1:", round(f1_scores[best_index], 4))

print("\nTHRESHOLD COMPARISON")
print("=" * 65)

print(threshold_results.round(4).to_string(index=False))
```

```python

# ==========================================
# STEP 10.5A: FINAL TEST SET EVALUATION
# ==========================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.metrics import (
    confusion_matrix,
    classification_report,
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    ConfusionMatrixDisplay,
    RocCurveDisplay,
    PrecisionRecallDisplay
)

# 1. Select final optimized model

final_model = best_wgb

# 2. Generate test probabilities

y_test_prob = final_model.predict_proba(X_test)[:, 1]

# 3. Apply optimized threshold

final_threshold = optimal_threshold

y_test_pred = (
    y_test_prob >= final_threshold
).astype(int)

# 4. Calculate final metrics

final_results = {
    "Accuracy": accuracy_score(y_test, y_test_pred),
    "Precision": precision_score(y_test, y_test_pred),
    "Recall": recall_score(y_test, y_test_pred),
    "F1-Score": f1_score(y_test, y_test_pred),
    "ROC-AUC": roc_auc_score(y_test, y_test_prob),
    "PR-AUC": average_precision_score(y_test, y_test_prob)
}

# 5. Display final metrics

print("=" * 55)
print("FINAL TEST SET PERFORMANCE")
print("=" * 55)

for metric, value in final_results.items():
    print(f"{metric}: {value:.4f}")


# 6. Classification report

print("\nCLASSIFICATION REPORT")
print("=" * 55)

print(classification_report(
    y_test,
    y_test_pred,
    target_names=["No Churn", "Churn"],
    digits=4
))


# 7. Confusion matrix

cm = confusion_matrix(
    y_test,
    y_test_pred,
    labels=[0, 1]
)

print("\nCONFUSION MATRIX")
print(cm)

disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["No Churn", "Churn"]
)

disp.plot(cmap="Blues", values_format="d")
plt.title("Final Model - Confusion Matrix")
plt.tight_layout()
plt.show()


# 8. ROC curve

RocCurveDisplay.from_predictions(
    y_test,
    y_test_prob,
    name="Weighted Gradient Boosting"
)

plt.plot(
    [0, 1],
    [0, 1],
    linestyle="--",
    label="Random Classifier"
)

plt.title("Final Model - ROC Curve")
plt.tight_layout()
plt.show()


# 9. Precision-Recall curve

PrecisionRecallDisplay.from_predictions(
    y_test,
    y_test_prob,
    name="Weighted Gradient Boosting"
)

plt.title("Final Model - Precision-Recall Curve")
plt.tight_layout()
plt.show()


# 10. Save final evaluation results

final_results_df = pd.DataFrame(
    [final_results]
)

print("\nFINAL EVALUATION COMPLETED")
```

```python
import pandas as pd
import joblib

# Load the saved artifact
artifact = joblib.load("customer_churn_model.joblib")

loaded_model = artifact["model"]
threshold = artifact["threshold"]
required_features = artifact["features"]


def predict_customer_churn(customer_data):

    # Validate input
    missing_features = [
        col for col in required_features
        if col not in customer_data.columns
    ]

    if missing_features:
        raise ValueError(
            f"Missing required features: {missing_features}"
        )

    # Select features in the correct order
    customer_data = customer_data[required_features]

    # Predict churn probabilities
    probabilities = loaded_model.predict_proba(customer_data)[:, 1]

    # Apply optimized threshold
    predictions = (probabilities >= threshold).astype(int)

    # Create results
    results = pd.DataFrame({
        "Churn Probability": probabilities,
        "Churn Prediction": predictions,
        "Churn Label": [
            "Churn" if pred == 1 else "No Churn"
            for pred in predictions
        ]
    }, index=customer_data.index)

    return results
```
